## 1 · Pedidos por estado
¿Cuántos pedidos hay en cada estado (o_orderstatus) y cuál es el monto total de cada uno?

Pista: groupBy sobre orders con dos agregaciones: count y sum de o_totalprice.

### Identificar primero los diferentes status que tienen las ordenes por si se debe realizar limpieza

```python
df_StatusUnique = df_orders.select("o_orderstatus").distinct()
display(df_StatusUnique)
```

### Se realiza la agrupación por el campo o_orderstatus se suman los montos y se cuenta las filas

```python

df_OrderByStatus =  df_orders.groupBy("o_orderstatus").agg(
    F.sum("o_totalprice").alias("total_ventas"), F.count("*").alias("cantidad")
)

display(df_OrderByStatus)
```

### Comprobacion para un estado dado

```python
display(df_orders.filter("o_orderstatus = 'O' " ).count())
df_orders.filter("o_orderstatus = 'O' " ).agg(
    F.sum("o_totalprice").alias("total_ventas")).show()
```
